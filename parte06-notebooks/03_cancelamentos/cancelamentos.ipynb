# análise de dataset

A empresa Umbrella Corporation contratou desenvolvedores python para analisar um dataset um dataset para verificar o alto número de cancelamentos de contratos dessa empresa, e para entregar uma solução para este problema.

para isso, a Umbrella Corp. forneceu um *dataset* com os detalhes dos cancelamentos. Os Pythonianos (🐍) devem analisar o arquivo utilizando o **pandas** (🐻‍❄️).

Vamos começar trazendo todos os dados do *dataset*:

```python
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")
display(tabela)
```

## Tratamento de dados 

Ao ler o *dataset*, percebemos que existem células com valores nulos, que atrapalham a nossa análise, e portanto devem ser retirados da análise.

Vamos elimina-lo e re-exibir os valores:

```python
tabela = tabela.drop("CustomeID", axis=1)
display(tabela)
```

Vamos pedir para que o Python nos retorne o número de linhas preenchidas da tabela.


```python
display(tabela.info())
```

Veja que temos em cada uma das colunas um número diferente de linhas. Issi acontece porque temos dados nulos na tabela, e que devemos eliminar, portanto, vamos eliminá-los através do comando python abaixo e re-exibir a tabela: 

```python
tabela = tabela.dropna()
display(tabela.info())
```

## Análise de dados 

Agora sim é começa a análise de dados propriamente dita.
Vamos começar verificando o número de clientes que cancelaram o contrato e o número de clientes que não cancelaram.

```python
display(tabela["cancelou"].value_counts())
```

Podemos perceber que o número de clientes que cancelaram (1.0)
realmente é alto, inclusive mais alto que o número de clientes que não cancelaram, mas quanto que isso dá em porcentagem? Vmos decobrir.

```python
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))
```

Sabemos agora que aproximadamente 56% dos clientes cancelaram seus contrato. Nosso trabalho é descobriro porquê.

Vamos analisar a taxa de cancelamentos por duração de contrato:

```python
display(tabela["duracao_contrato"].value_counts())
display(tabela["duracao_contrato"].value_counts(normalize=True).map("{:.2%}".format))
```

Podemos perceber que existe uma minoria em relação aos contratos mensais, o que talvez possa significar que os cancelamentos estejam vindo dos contratos mensais.

o que acontece se retirarmos da análise os contratos mensais? Vamos análisar a taxa de cancelamentos por contrato:

```python
display(tabela.groupby("duracao_contrato").mean(numeric_only=True))
```

Logo de cara, percebemos que todos os clientes que optaram pel tipo de contrato mensal cancelaram o serviço.


## o que pode ser feito para resolver o problema dos contratos mensais?

podems retirar o tipo de contrato mensal da análise para verificar qual a porcentagem de cancelamentos que sobra:

```python
tabela = tabela[tabela["duracao_contrato"] != "Monthly"]
display(tabela)
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2}".format))
```

Podemos verificar que sem o contrato do tipoi mensal, a taxa de cancelamentos diminuiu para 46%. só que a taxa de cancelamentos continua muito alta. Portanto, precisaremos fazer uma análise ainda mais profunda nos contratos do tipo anual e trimestral.

Para termos uma noção melhor, vamos criar gráficos de cada coluna  para verificar se há discrepâncias muito grandes nas taxas de cancelamentos em alguma dessas colunas:

```python
import plotly.express as px

for coluna in tabela.columns:
    grafico = px.histogram(tabela, x=coluna,color="cancelou", width=600)

    grafico.show()
```

Os gráficos indicam problemas com os usuários que fazem mais que 5 ligações para o call center, e também cancelamentos por parte dos usuários que passam mais de 20 dias inadiplentes. portanto, devemos eliminar esses clientes da análise.

```python
tabela =tabela[tabela["ligacoes_callcenter"] < 5]
tabela = tabela[tabela["dias_atraso"] <= 20]
display(tabela)
display(tabela["cancelou"].value_counts())
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2f}".format))
```
